# Measuring Foreign Exposure: Synchronization

This notebook contains `Stata 17 MP 8-core` code that produces Table 3. Imbs & Pauwels, *Measuring Foreign Exposure*, JIE, 2025.  Package requirements for the code are:
* `reghdfe`
* `ftools`
* `estout`
* `outreg2`

These packages can be installed in Stata with the `ssc install` command. It also uses custom **.do** files: `loadBilateralData.do`, `processSynchronization.do`, and `makeTablesSynchronization.do` available from the repo.

### Working directory name

```stata
clear 
set more off
global dir `c(pwd)'
```

### Run `loadBilateralData.do.do` and `processSynchronization.do` 

`loadBilateralData.do` loads the data `bilateral.txt` produced my MATLAB in `matlab/output`.

`processSynchronization.do` prepares the variables required for Table 4. This includes removing the Rest of the World (ROW) from the data, and taking logs of the openness measures, namely HOT, TiVA, Exports, and "Phi". It also defines three aggregate sectors: Agriculture (AGR), Manufacturing (MFG), and Services (SER).

<span style="color:red;"> Warning: Loading the data and preparing the data has a runtime of about 5 minutes. Save your work if you do not wish to run this every time you clear Stata's memory. </span>

```stata
quietly{
    do "./scripts/loadBilateralData.do"
    cd ../../stata
    do "./scripts/processSynchronization.do"
}
describe
```
